```python
import sys
import argparse
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain.chat_models import ChatOpenAI
from langchain.prompts import ChatPromptTemplate

CHROMA_PATH = "chroma"

PROMPT_TEMPLATE = """"
    Answer the question based only on the following context:

    {context}

    --------------------------------
    Answer the question based on the above context: {question}
"""

def chatbot():
    print("\n🤖 **Welcome to the AI Chatbot!** Type 'exit' or 'quit' to stop.\n")

    # Initialize embeddings and database
    embedding_function = OpenAIEmbeddings()
    db = Chroma(persist_directory=CHROMA_PATH, embedding_function=embedding_function)

    # Initialize LLM model
    model = ChatOpenAI()

    while True:
        # Get user input
        query_text = print(input("\nYou: "))

        # Exit conditions
        if query_text.lower() in ["exit", "quit"]:
            print("\n🤖 Chatbot: Goodbye! Have a great day! 🚀")
            break

        # Retrieve most relevant documents
        results = db.similarity_search_with_relevance_scores(query_text, k=3)
        
        # Handle no relevant results
        if not results or results[0][1] < 0.7:
            print("\n🤖 Chatbot: Sorry, I couldn't find a good match. Can you rephrase?")
            continue

        # Format the retrieved content
        context_text = "\n\n------\n\n".join([doc.page_content for doc, _ in results])
        prompt_template = ChatPromptTemplate.from_template(PROMPT_TEMPLATE)
        prompt = prompt_template.format(context=context_text, question=query_text)

        # Generate response
        response_text = model.predict(prompt)

        # Get sources
        sources = [doc.metadata.get("source", None) for doc, _ in results]

        # Display chatbot response
        print(f"\n🤖 Chatbot: {response_text}")
        

if __name__ == "__main__":
    chatbot()  # Start chatbot loop
```
